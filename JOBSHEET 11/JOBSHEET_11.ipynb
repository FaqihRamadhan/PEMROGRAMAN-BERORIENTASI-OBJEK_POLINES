```python
# Kelas Induk Utama
class Kendaraan:
    def __init__(self, merk: str):
        self.merk = merk

    def info_kendaraan(self):
        print(f"Kendaraan Merk: {self.merk}")


# Kelas Anak (Mewarisi Kendaraan)
class Mobil(Kendaraan):
    def __init__(self, merk: str, model: str):
        # Melewatkan argumen ke konstruktor kelas induk
        super().__init__(merk)
        self.model = model

    def bergerak(self):
        print(f"Mobil {self.merk} model {self.model} sedang bergerak maju.")


# Kelas Cucu (Mewarisi Mobil) - Pewarisan Multi-level
class MobilListrik(Mobil):
    def __init__(self, merk: str, model: str, kapasitas_baterai: int):
        # Melewatkan argumen ke konstruktor kelas Mobil
        super().__init__(merk, model)
        self.kapasitas_baterai = kapasitas_baterai

    def isi_daya(self):
        print(f"Mengisi daya Mobil Listrik {self.merk} {self.model}. Kapasitas Baterai: {self.kapasitas_baterai} kWh.")


# --- DRIVER CODE (Demonstrasi Objek & Pemanggilan Method) ---
if __name__ == "__main__":
    print("=== DEMONSTRASI BAGIAN A (KENDARAAN) ===")
    
    # Instansiasi objek MobilListrik (konkrit paling bawah)
    mobil_tesla = MobilListrik("Tesla", "Model Y", 75)
    
    # Pemanggilan method yang diwarisi dan milik sendiri
    mobil_tesla.info_kendaraan()  # Warisan dari Kendaraan
    mobil_tesla.bergerak()        # Warisan dari Mobil
    mobil_tesla.isi_daya()        # Method asli milik MobilListrik
    print()
```
