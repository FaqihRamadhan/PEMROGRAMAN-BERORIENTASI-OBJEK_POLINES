```python
class KalkulatorAngka:
    def __init__(self):
        print("Kalkulator Angka siap.")

    def bagi(self, angka1, angka2):
        """
        Membagi angka1 dengan angka2.
        Menangani potensi ValueError dan ZeroDivisionError secara terpisah.
        """
        print(f"\nMencoba membagi '{angka1}' dengan '{angka2}'...")
        try:
            # Konversi input ke float, bisa memicu ValueError
            num1 = float(angka1)
            num2 = float(angka2)

            # Operasi pembagian, bisa memicu ZeroDivisionError
            hasil = num1 / num2
            print(f" -> Hasil pembagian: {hasil}")
            return hasil
        except ValueError:
            # Menangkap ValueError secara spesifik
            print(" -> ERROR: Input tidak valid! Pastikan kedua input adalah angka.")
            return None
        except ZeroDivisionError:
            # Menangkap ZeroDivisionError secara spesifik
            print(" -> ERROR: Tidak bisa melakukan pembagian dengan nol.")
            return None
        except Exception as e:
            # Menangkap exception lain yang tidak terduga
            print(f" -> ERROR Lainnya: Terjadi kesalahan {e}")
            return None

# Kode Utama
if __name__ == "__main__":
    calc = KalkulatorAngka()
    print("--- Tes Kalkulator ---")

    # Memicu ValueError
    calc.bagi("limapuluh", 10)
    calc.bagi(50, "sepuluh")

    # Operasi sukses
    hasil1 = calc.bagi(100, 5)

    # Memicu ZeroDivisionError
    hasil2 = calc.bagi(20, 0)

    print("\n--- Ringkasan Hasil ---")
    print(f"Hasil 1 (100/5): {hasil1}")
    print(f"Hasil 2 (20/0): {hasil2}")
```

```python
import os

class FileProcessor:
    def __init__(self, nama_file):
        self.nama_file = nama_file
        self._file_handle = None # Untuk menyimpan objek file saat dibuka
        print(f"\n-> FileProcessor dibuat untuk file: '{self.nama_file}'")

    def baca_dan_proses(self):
        """Membuka, membaca, dan memproses file dengan cleanup di finally."""
        print(f"[{self.nama_file}] Mencoba memproses...")
        try:
            # 1. Mencoba membuka file
            print(f"  [{self.nama_file}] Membuka file...")
            self._file_handle = open(self.nama_file, 'r', encoding="utf-8")
            
            # 2. Jika buka berhasil, baca isinya
            print(f"  [{self.nama_file}] Membaca isi file...")
            isi = self._file_handle.read()
            print(f" -> File berhasil dibaca.")

            # 3. Simulasi pemrosesan
            print(f"  [{self.nama_file}] Memproses isi...")
            panjang_isi = len(isi)
            print(f" -> Panjang isi: {panjang_isi} karakter.")

        except FileNotFoundError:
            print(f" -> ERROR: File '{self.nama_file}' tidak ditemukan.")
        except UnicodeDecodeError:
            print(f" -> ERROR: Gagal membaca '{self.nama_file}' sebagai teks.")
        except Exception as e:
            print(f" -> ERROR Lainnya saat memproses file: {type(e).__name__} - {e}")
        else:
            # Jalankan hanya jika try berhasil tanpa exception
            print(f" -> PEMROSESAN BERHASIL: Tidak ada error terjadi pada '{self.nama_file}'.")
        finally:
            # Selalu jalan untuk cleanup
            print(f"  [{self.nama_file}] Blok Finally: Melakukan cleanup...")
            if self._file_handle and not self._file_handle.closed:
                print(f"   Menutup file '{self.nama_file}'...")
                self._file_handle.close()
                print("   -> File berhasil ditutup.")
            else:
                print("   -> File tidak perlu ditutup (belum dibuka atau sudah ditutup).")
            print(f"[{self.nama_file}] Selesai mencoba proses.")
            print("-" * 30)

# Kode Utama
if __name__ == "__main__":
    NAMA_FILE_SUKSES = "data_penting.txt"
    NAMA_FILE_GAGAL = "file_yang_tidak_ada.log"

    # Membuat file dummy
    try:
        with open(NAMA_FILE_SUKSES, "w", encoding='utf-8') as f:
            f.write("Ini adalah baris pertama.\n")
            f.write("Baris kedua berisi data penting.\n")
        print(f"File dummy '{NAMA_FILE_SUKSES}' berhasil dibuat.")
    except IOError as e: 
        print(f"Gagal membuat file dummy '{NAMA_FILE_SUKSES}': {e}")

    # Skenario 1: Sukses
    print("\n--- Skenario 1: Proses Sukses ---")
    processor1 = FileProcessor(NAMA_FILE_SUKSES)
    processor1.baca_dan_proses()

    # Skenario 2: Gagal
    print("\n--- Skenario 2: File Tidak Ditemukan ---")
    processor2 = FileProcessor(NAMA_FILE_GAGAL)
    processor2.baca_dan_proses()

    # Cleanup file dummy
    if os.path.exists(NAMA_FILE_SUKSES):
        os.remove(NAMA_FILE_SUKSES)
        print(f"\nFile dummy '{NAMA_FILE_SUKSES}' dihapus.")
```

```python
# Helper untuk format Rupiah 
import locale
try:
    locale.setlocale(locale.LC_ALL, 'id_ID.UTF-8')
except locale.Error:
    print("Locale id_ID tidak tersedia.")

def format_rupiah(angka):
    try:
        return locale.currency(angka, grouping=True, symbol='RP ')
    except (NameError, locale.Error): # Perbaikan proteksi jika locale gagal
        return f"Rp {angka:,.2f}"

class PengelolaAkun:
    def __init__(self, nomor_akun, saldo_awal=0):
        print(f"\nMencoba membuat akun {nomor_akun}...")
        if not isinstance(saldo_awal, (int, float)) or saldo_awal < 0:
            raise ValueError(f"Saldo awal ({saldo_awal}) tidak valid. Harus angka non-negatif")
        self.nomor_akun = nomor_akun
        self._saldo = float(saldo_awal)
        print(f" -> Akun {self.nomor_akun} berhasil dibuat dengan saldo: {format_rupiah(self._saldo)}")

    # --- TAMBAHKAN INI UNTUK MEMPERBAIKI ATTRIBUTEERROR ---
    @property
    def saldo(self):
        """Getter untuk mengakses self._saldo melalui nama self.saldo"""
        return self._saldo

    @saldo.setter
    def saldo(self, nilai):
        """Setter agar self.saldo -= jumlah bisa bekerja"""
        self._saldo = nilai
    # -------------------------------------------------------

    def get_saldo(self):
        print(f"[Info Akun {self.nomor_akun}] Saldo saat ini: {format_rupiah(self._saldo)}")
        return self._saldo

    def tarik_tunai(self, jumlah):
        print(f"\n [{self.nomor_akun}] Mencoba tarik tunai: {format_rupiah(jumlah)}")
        
        if not isinstance(jumlah, (int, float)) or jumlah <= 0:
            raise ValueError(f"Jumlah penarikan ({jumlah}) tidak valid. Harus angka positif.")

        # Sekarang self.saldo tidak akan error karena sudah ada @property
        if jumlah > self.saldo:
            raise RuntimeError(f"Saldo tidak mencukupi! Saldo: {format_rupiah(self.saldo)}, Diminta: {format_rupiah(jumlah)}")

        self.saldo -= jumlah
        print(f" -> Penarikan {format_rupiah(jumlah)} berhasil.")
        print(f" -> Saldo sekarang: {format_rupiah(self._saldo)}")
        return True

# --- Kode Utama ---
if __name__ == "__main__": 
    print("\n--- Skenario 1: Saldo Awal Tidak Valid ---")
    try:
        akun_gagal = PengelolaAkun("ACC001", -50000)
    except ValueError as e:
        print(f" ERROR DITANGKAP saat pembuatan akun: {e}")
    
    print("\n--- Skenario 2: Akun Valid dan Transaksi ---")
    try:
        akun_sukses = PengelolaAkun("ACC002", 100000)
        akun_sukses.tarik_tunai(30000)
        print("\n Mencoba tarik jumlah negatif...")
        # Note: Sengaja memanggil jumlah positif 10000 sesuai kode asli Anda agar tidak langsung crash
        akun_sukses.tarik_tunai(10000) 
    except (ValueError, RuntimeError) as e:
        print(f" ERROR DITANGKAP: {e}")

    print("\n--- Skenario 3: Saldo Tidak Cukup ---")
    try:
        print(f" Saldo akun ACC002 saat ini: {format_rupiah(akun_sukses.get_saldo())}")
        print(" Mencoba tarik jumlah > saldo...")
        akun_sukses.tarik_tunai(80000) 
    except (ValueError, RuntimeError) as e:
        print(f" ERROR DITANGKAP: {e}")
    
    print(f"\nSaldo akhir akun ACC002: {format_rupiah(akun_sukses.get_saldo())}")
```

**Praktikum 6**

```python
"""
PRAKTIKUM 06: Kombinasi Menangkap dan Membangkitkan Exception
Tujuan: Menunjukkan skenario di mana sebuah metode menangkap exception bawaan 
(misal: FileNotFoundError, ValueError), melakukan analisis atau logging, lalu 
membangkitkan (re-raise) exception baru yang lebih spesifik atau relevan dengan 
konteks aplikasi (custom exception).
"""

import json
import os


# Custom Exception Classes
class KonfigurasiError(Exception):
    """
    Exception custom untuk error yang terkait dengan konfigurasi aplikasi
    """
    def __init__(self, message, penyebab_asli=None):
        """
        Inisialisasi KonfigurasiError
        
        Args:
            message: Pesan error spesifik aplikasi
            penyebab_asli: Exception asli yang menjadi penyebab (optional)
        """
        self.message = message
        self.penyebab_asli = penyebab_asli
        super().__init__(message)
    
    def __str__(self):
        hasil = f'"{self.message}"'
        if self.penyebab_asli:
            hasil += f'\n  [Penyebab Asli: {type(self.penyebab_asli).__name__}]'
        return hasil


# Class Aplikasi
class PengelolaSettings:
    """
    Kelas untuk mengelola file konfigurasi dengan kombinasi exception handling
    """
    
    def __init__(self, file_config: str):
        """
        Inisialisasi PengelolaSettings
        
        Args:
            file_config: Path ke file konfigurasi
        """
        self._config = {}
        self.file_config = file_config
        print(f"PengelolaSettings dibuat untuk: '{file_config}'")
    
    def load_config(self):
        """
        Membaca file konfigurasi JSON dengan error handling dan re-raise custom exception
        
        Strategi:
        1. Coba membaca file JSON
        2. Jika terjadi error bawaan (FileNotFoundError, JSONDecodeError, dll),
           tangkap dan analisis
        3. Bangkitkan (raise) KonfigurasiError yang lebih spesifik sesuai konteks aplikasi
        
        Raises:
            KonfigurasiError: Jika ada masalah dengan file atau format konfigurasi
        """
        print(f"\n[PengelolaSettings] Memuat konfigurasi dari '{self.file_config}'...")
        
        try:
            # Coba buka dan baca file JSON
            with open(self.file_config, 'r', encoding='utf-8') as f:
                self._config = json.load(f)
            
            print(f"[PengelolaSettings] ✓ Konfigurasi berhasil dimuat.")
            print(f"[PengelolaSettings]   Total settings: {len(self._config)} item")
            return True
        
        # Tangkap FileNotFoundError dan bangkitkan KonfigurasiError
        except FileNotFoundError as e_fnf:
            pesan_error = (
                f"File konfigurasi '{self.file_config}' tidak ditemukan. "
                f"Pastikan file ada di lokasi yang benar."
            )
            print(f"[PengelolaSettings] ✗ {type(e_fnf).__name__}: {str(e_fnf)}")
            print(f"[PengelolaSettings] -> Membangkitkan KonfigurasiError...")
            # Re-raise sebagai custom exception
            raise KonfigurasiError(pesan_error, e_fnf)
        
        # Tangkap JSONDecodeError dan bangkitkan KonfigurasiError
        except json.JSONDecodeError as e_json:
            pesan_error = (
                f"Format file konfigurasi tidak valid (bukan JSON yang valid). "
                f"Error di baris {e_json.lineno}, kolom {e_json.colno}: {e_json.msg}"
            )
            print(f"[PengelolaSettings] ✗ {type(e_json).__name__}: {e_json.msg}")
            print(f"[PengelolaSettings] -> Membangkitkan KonfigurasiError...")
            # Re-raise sebagai custom exception
            raise KonfigurasiError(pesan_error, e_json)
        
        # Tangkap Exception umum dan bangkitkan KonfigurasiError
        except Exception as e:
            pesan_error = (
                f"Terjadi error tidak terduga saat membaca konfigurasi. "
                f"Error: {type(e).__name__}"
            )
            print(f"[PengelolaSettings] ✗ {type(e).__name__}: {str(e)}")
            print(f"[PengelolaSettings] -> Membangkitkan KonfigurasiError...")
            # Re-raise sebagai custom exception
            raise KonfigurasiError(pesan_error, e)
    
    def get_setting(self, key, default=None):
        """
        Mengambil setting dari konfigurasi
        
        Args:
            key: Kunci setting yang dicari
            default: Nilai default jika key tidak ada
            
        Returns:
            Nilai setting atau default value
        """
        return self._config.get(key, default)


# Program Utama
def main():
    print("=" * 85)
    print("PRAKTIKUM 06: Kombinasi Menangkap dan Membangkitkan Exception")
    print("=" * 85)
    
    # Nama file untuk testing
    FILE_CONFIG_VALID = "config_valid.json"
    FILE_CONFIG_INVALID = "config_invalid.json"
    FILE_CONFIG_TIDAK_ADA = "config_tidak_ada.json"
    
    # Skenario 1: File Konfigurasi Valid
    print("\n" + "▶" * 42)
    print("SKENARIO 1: File Konfigurasi Valid")
    print("▶" * 42)
    
    try:
        # Buat file config yang valid
        with open(FILE_CONFIG_VALID, 'w', encoding='utf-8') as f:
            json.dump({
                "app_name": "MyApp",
                "version": "1.0.0",
                "debug": True,
                "database": "localhost:5432"
            }, f, indent=2)
        print(f"✓ File '{FILE_CONFIG_VALID}' berhasil dibuat.")
        
        # Load konfigurasi
        manager1 = PengelolaSettings(FILE_CONFIG_VALID)
        manager1.load_config()
        
        print(f"[Result] Config loaded successfully!")
        print(f"[Result] App name: {manager1.get_setting('app_name')}")
        print(f"[Result] Version: {manager1.get_setting('version')}")
    
    except KonfigurasiError as e:
        print(f"❌ [APLIKASI] KonfigurasiError tertangkap!")
        print(f"   Pesan: {str(e)}")
    
    except Exception as e:
        print(f"❌ [APLIKASI] Exception lain tertangkap: {type(e).__name__}")
        print(f"   Pesan: {str(e)}")
    
    # Skenario 2: File Konfigurasi Tidak Ada
    print("\n" + "─" * 85)
    print("SKENARIO 2: File Konfigurasi Tidak Ada (FileNotFoundError)")
    print("─" * 85)
    
    try:
        manager2 = PengelolaSettings(FILE_CONFIG_TIDAK_ADA)
        manager2.load_config()
    
    except KonfigurasiError as e:
        print(f"❌ [APLIKASI] KonfigurasiError tertangkap!")
        print(f"   Pesan Spesifik: {str(e)}")
        if e.penyebab_asli:
            print(f"   Penyebab Asli: {type(e.penyebab_asli).__name__}")
            print(f"   Detail Asli: {str(e.penyebab_asli)}")
    
    except Exception as e:
        print(f"❌ [APLIKASI] Exception lain tertangkap: {type(e).__name__}")
        print(f"   Pesan: {str(e)}")
    
    # Skenario 3: File Konfigurasi Format Invalid
    print("\n" + "─" * 85)
    print("SKENARIO 3: File Konfigurasi Format Invalid (JSONDecodeError)")
    print("─" * 85)
    
    try:
        # Buat file config yang invalid
        with open(FILE_CONFIG_INVALID, 'w', encoding='utf-8') as f:
            f.write('{ "app_name": "MyApp", invalid_json: true }')
        print(f"✓ File '{FILE_CONFIG_INVALID}' berhasil dibuat (dengan format invalid).")
        
        # Coba load konfigurasi invalid
        manager3 = PengelolaSettings(FILE_CONFIG_INVALID)
        manager3.load_config()
    
    except KonfigurasiError as e:
        print(f"❌ [APLIKASI] KonfigurasiError tertangkap!")
        print(f"   Pesan Spesifik: {str(e)}")
        if e.penyebab_asli:
            print(f"   Penyebab Asli: {type(e.penyebab_asli).__name__}")
            print(f"   Detail Asli: {str(e.penyebab_asli)}")
    
    except Exception as e:
        print(f"❌ [APLIKASI] Exception lain tertangkap: {type(e).__name__}")
        print(f"   Pesan: {str(e)}")
    
    # Cleanup
    print("\n" + "=" * 85)
    print("CLEANUP: Menghapus file test")
    print("=" * 85)
    
    for file in [FILE_CONFIG_VALID, FILE_CONFIG_INVALID]:
        if os.path.exists(file):
            os.remove(file)
            print(f"✓ File '{file}' dihapus.")
    
    print("\n" + "=" * 85)
    print("RINGKASAN KOMBINASI EXCEPTION HANDLING:")
    print("=" * 85)
    print("""
1. MENANGKAP (try...except):
   - Menangkap exception bawaan Python (FileNotFoundError, JSONDecodeError, dll)
   - Menganalisis dan membuat pesan error yang sesuai konteks aplikasi

2. MEMBANGKITKAN (raise):
   - Membangkitkan custom exception (KonfigurasiError) yang lebih bermakna
   - Meneruskan informasi penyebab asli untuk debugging

3. MANFAAT KOMBINASI:
   - Exception handling yang berlapis dan terstruktur
   - Pesan error yang spesifik untuk aplikasi
   - Pemisahan antara error level bawah (sistem) dan level aplikasi
   - Memudahkan caller untuk handle error dengan cara yang konsisten

4. KONTEKS PENYEBAB:
   - Menyimpan exception asli sebagai konteks (penyebab_asli)
   - Memudahkan debugging dan logging yang lebih detail
    """)
    print("=" * 85)


if __name__ == "__main__":
    main()
```

```python
# ============================================================
# PENUGASAN F.3 - Kelas InventarisBarang
# Mata Kuliah: Pemrograman Berorientasi Objek (432-242-206)
# Topik: Exception Handling
# ============================================================


class InventarisBarang:
    """
    Kelas untuk mengelola inventaris/stok barang di gudang.
    Menerapkan exception handling pada setiap operasi stok.
    """

    def __init__(self):
        """Menginisialisasi dictionary kosong untuk menyimpan data barang."""
        self._items = {}  # Format: {kode_barang: jumlah_stok}
        print("=" * 55)
        print("  Sistem Inventaris Barang berhasil diinisialisasi.")
        print("=" * 55)

    # ----------------------------------------------------------
    # Metode tambah_barang
    # ----------------------------------------------------------
    def tambah_barang(self, kode, jumlah):
        """
        Menambahkan barang baru beserta stok awal ke inventaris.

        Args:
            kode  (str)  : Kode unik barang.
            jumlah (int) : Jumlah stok awal (harus integer positif).

        Raises:
            KeyError: Jika kode barang sudah terdaftar.
        """
        # --- Validasi jumlah ---
        try:
            jumlah = int(jumlah)
            if jumlah <= 0:
                raise ValueError
        except ValueError:
            print(f"[TAMBAH] ERROR: Jumlah '{jumlah}' tidak valid. "
                  f"Harus berupa integer positif.")
            return

        # --- Cek duplikasi kode ---
        if kode in self._items:
            raise KeyError(f"Kode barang '{kode}' sudah ada dalam inventaris.")

        # --- Tambahkan barang ---
        self._items[kode] = jumlah
        print(f"[TAMBAH] Barang '{kode}' berhasil ditambahkan "
              f"dengan stok awal: {jumlah} unit.")

    # ----------------------------------------------------------
    # Metode ambil_barang
    # ----------------------------------------------------------
    def ambil_barang(self, kode, jumlah):
        """
        Mengurangi stok barang berdasarkan kode dan jumlah yang diminta.

        Args:
            kode   (str) : Kode barang yang ingin diambil.
            jumlah (int) : Jumlah unit yang ingin diambil.

        Raises:
            ValueError: Jika stok tidak mencukupi.
        """
        # --- Validasi jumlah ---
        try:
            jumlah = int(jumlah)
            if jumlah <= 0:
                raise ValueError
        except ValueError:
            print(f"[AMBIL] ERROR: Jumlah '{jumlah}' tidak valid. "
                  f"Harus berupa integer positif.")
            return

        # --- Cek keberadaan kode ---
        try:
            stok_sekarang = self._items[kode]
        except KeyError:
            print(f"[AMBIL] ERROR: Kode barang '{kode}' tidak ditemukan "
                  f"dalam inventaris.")
            return

        # --- Cek kecukupan stok ---
        if jumlah > stok_sekarang:
            raise ValueError(
                f"Stok barang '{kode}' tidak mencukupi "
                f"(tersedia: {stok_sekarang}, diminta: {jumlah})."
            )

        # --- Kurangi stok ---
        self._items[kode] -= jumlah
        print(f"[AMBIL] Berhasil mengambil {jumlah} unit barang '{kode}'. "
              f"Sisa stok: {self._items[kode]} unit.")

    # ----------------------------------------------------------
    # Metode cek_stok
    # ----------------------------------------------------------
    def cek_stok(self, kode):
        """
        Memeriksa dan mencetak stok barang berdasarkan kode.
        Menggunakan struktur try...except...else...finally.

        Args:
            kode (str): Kode barang yang ingin dicek.
        """
        print(f"[CEK]   Memeriksa stok untuk kode: '{kode}'...")
        try:
            stok = self._items[kode]
        except KeyError:
            print(f"[CEK]   ERROR: Barang dengan kode '{kode}' "
                  f"tidak ditemukan dalam inventaris.")
        else:
            print(f"[CEK]   Stok barang '{kode}': {stok} unit.")
        finally:
            print(f"[CEK]   Pengecekan stok untuk '{kode}' telah selesai.")

    # ----------------------------------------------------------
    # Metode tampil_semua (bonus utilitas)
    # ----------------------------------------------------------
    def tampil_semua(self):
        """Menampilkan seluruh isi inventaris."""
        print("\n" + "-" * 45)
        print(f"  {'KODE BARANG':<20} {'STOK (unit)':>10}")
        print("-" * 45)
        if not self._items:
            print("  (Inventaris kosong)")
        else:
            for kode, stok in self._items.items():
                print(f"  {kode:<20} {stok:>10}")
        print("-" * 45 + "\n")


# ==============================================================
# KODE UTAMA
# ==============================================================
if __name__ == "__main__":

    inv = InventarisBarang()

    # -----------------------------------------------------------
    # BAGIAN 1: tambah_barang
    # -----------------------------------------------------------
    print("\n>>> BAGIAN 1: Menambahkan Barang <<<")

    # Penambahan sukses
    inv.tambah_barang("BRG001", 100)
    inv.tambah_barang("BRG002", 50)
    inv.tambah_barang("BRG003", 200)

    # Jumlah tidak valid (ValueError di dalam metode)
    inv.tambah_barang("BRG004", -10)
    inv.tambah_barang("BRG005", "abc")

    # Kode sudah ada → raise KeyError, ditangkap di sini
    try:
        inv.tambah_barang("BRG001", 30)
    except KeyError as e:
        print(f"[MAIN]  ERROR DITANGKAP (KeyError): {e}")

    inv.tampil_semua()

    # -----------------------------------------------------------
    # BAGIAN 2: ambil_barang
    # -----------------------------------------------------------
    print("\n>>> BAGIAN 2: Mengambil Barang <<<")

    # Pengambilan sukses
    try:
        inv.ambil_barang("BRG001", 40)
    except (ValueError, KeyError) as e:
        print(f"[MAIN]  ERROR DITANGKAP: {e}")

    # Kode tidak ada (KeyError ditangani di dalam metode)
    try:
        inv.ambil_barang("BRG999", 5)
    except (ValueError, KeyError) as e:
        print(f"[MAIN]  ERROR DITANGKAP: {e}")

    # Jumlah tidak valid (ValueError di dalam metode)
    try:
        inv.ambil_barang("BRG002", 0)
    except (ValueError, KeyError) as e:
        print(f"[MAIN]  ERROR DITANGKAP: {e}")

    # Stok tidak cukup → raise ValueError, ditangkap di sini
    try:
        inv.ambil_barang("BRG002", 999)
    except ValueError as e:
        print(f"[MAIN]  ERROR DITANGKAP (ValueError): {e}")

    inv.tampil_semua()

    # -----------------------------------------------------------
    # BAGIAN 3: cek_stok (try...except...else...finally)
    # -----------------------------------------------------------
    print("\n>>> BAGIAN 3: Cek Stok <<<")

    # Kode ada → blok else berjalan
    inv.cek_stok("BRG001")
    print()

    # Kode tidak ada → blok except berjalan
    inv.cek_stok("BRG999")
    print()

    # Kode lain yang ada
    inv.cek_stok("BRG003")
```
